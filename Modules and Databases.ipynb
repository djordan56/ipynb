```python
from sqlalchemy import create_engine, Table, Column, Integer, String, MetaData, select

# Step 1: Connect to SQLite database (creates books.db if it doesn't exist)
engine = create_engine('sqlite:///books.db', echo=False)
metadata = MetaData()

# Step 2: Define 'books' table (create if missing)
books_table = Table('books', metadata,
                    Column('id', Integer, primary_key=True, autoincrement=True),
                    Column('title', String),
                    Column('author', String),
                    Column('year', Integer)  # can be None
                   )
metadata.create_all(engine)  # creates the table if it doesn't exist

# Step 3: Insert data if table is empty
with engine.connect() as conn:
    result = conn.execute(select(books_table))
    if result.first() is None:
        conn.execute(books_table.insert(), [
            {'title': 'The Hobbit', 'author': 'J R R Tolkien', 'year': None},
            {'title': 'Eats, Shoots & Leaves', 'author': 'Lynne Truss', 'year': None}
        ])

# Step 4: Query and print titles alphabetically (16.8)
stmt = select(books_table.c.title).order_by(books_table.c.title)
with engine.connect() as conn:
    result = conn.execute(stmt)
    print("Book Titles (alphabetical order):")
    for row in result:
        print(row.title)
```
