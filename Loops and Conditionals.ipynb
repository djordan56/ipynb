```python
secrect = 9
guess = 3

if guess < secrect:
    print("too low")
elif guess > secrect:
    print("too high")
else:
    print("just right")
```

```python
small = True
green = True

if small and green:
    print("pea")
else:
    if small and not green:
        print("cherry")
    else:
        if not small and green:
            print("watermelon")
        else:
            print("pumpkin")
    
```

```python
guess_me = 7
number = 1 

while number <= guess_me:
    if number < guess_me:
        print("too low")
    elif number == guess_me:
        print("found it!")
        break
    else:
        print("oops")
        break
    number += 1
```

```python
guess_me = 5 

for number in range(10):
    if number < guess_me:
        print("too low")
    elif number == guess_me:
        print("found it!")
        break
    else:
        print("oops")
        break
```
