## 13.1

This section writes the current date as a string to a file named today.txt using Python's datetime module. 
The date is formatted as month/day/year and saved to the file.

```python
from datetime import date

# Get today's date
today = date.today()

# Format as M/D/YYYY (works on Windows)
today_string = today.strftime("%m/%d/%Y")

# Write the date to today.txt
with open("today.txt", "w") as file:
    file.write(today_string)

today_string
```

## 13.2

This section reads the contents of today.txt and stores the value into the variable today_string.

```python
with open("today.txt", "r") as file:
    today_string = file.read()

today_string
```

## 13.3

This section parses the string stored in today_string and converts it back into a date object.

```python
from datetime import datetime

parsed_date = datetime.strptime(today_string, "%m/%d/%Y").date()

parsed_date
```

## 15.1

This section uses the multiprocessing module to create three separate processes. 
Each process waits for a random number of seconds between 0 and 1, prints the current time, and then exits.

```python
import multiprocessing
import random
import time
from datetime import datetime

def worker(number):
    wait_time = random.uniform(0, 1)
    time.sleep(wait_time)
    return f"Process {number} finished at {datetime.now().strftime('%H:%M:%S.%f')}"

if __name__ == "__main__":
    with multiprocessing.Pool(3) as pool:
        results = pool.map(worker, [1, 2, 3])

    for result in results:
        print(result)

    print("All processes completed.")
```
