```python
class Solution:
    def sort012(self, arr):
        count0 = arr.count(0)
        count1 = arr.count(1)
        count2 = arr.count(2)

        arr[:] = [0]*count0 + [1]*count1 + [2]*count2
```

```python
class Solution:
    def binarysearch(self, arr, k):
        left = 0
        right = len(arr) - 1
        result = -1  

        while left <= right:
            mid = left + (right - left) // 2  

            if arr[mid] == k:
                result = mid      
                right = mid - 1   
            elif arr[mid] < k:
                left = mid + 1   
            else:
                right = mid - 1   

        return result
```
